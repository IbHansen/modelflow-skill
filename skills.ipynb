# ModelFlow plugins

Renders every skill (`SKILL.md`, without frontmatter) and its reference files, plugin by plugin.
Run from the repo root.

```python
from pathlib import Path
from IPython.display import Markdown, display

PLUGINS = Path.cwd() / 'plugins'

def body(path):
    """Text of a markdown file without a leading --- frontmatter block."""
    text = path.read_text(encoding='utf8')
    if text.startswith('---'):
        text = text.split('---', 2)[2]
    return text

def show_plugin(name, references=True):
    """Render all skills of one plugin, optionally with their reference files."""
    for skill in sorted((PLUGINS / name / 'skills').iterdir()):
        display(Markdown(f'# `{skill.name}`\n' + body(skill / 'SKILL.md')))
        if references:
            for ref in sorted(skill.glob('references/*.md')):
                display(Markdown(f'## reference: `{ref.name}`\n' + body(ref)))
    for agent in sorted((PLUGINS / name).glob('agents/*.md')):
        display(Markdown(f'# agent `{agent.stem}`\n' + body(agent)))
```

```python
show_plugin('modelflow-user')
```

```python
show_plugin('modelflow-modeller')
```

```python
show_plugin('modelflow-dev')
```
